# Day 1: Introduction to Programming & Python

## What is a programming language?
A **programming language** is a formal set of words, symbols and rules that lets us give instructions to a computer. Computers only understand machine code (binary: 0s and 1s), so programming languages act as a bridge: we write human-readable instructions, and a **compiler** or **interpreter** translates them into something the machine can execute.

Programming languages are used to build almost everything around us: websites, mobile apps, games, banking systems, medical software, robots and AI models.

**Common types of programming languages**

| Type | Idea | Examples |
|------|------|----------|
| Low-level | Close to the hardware, fast but hard to read | Assembly, Machine code |
| High-level | Close to human language, easier to learn and write | Python, Java, JavaScript, C# |
| Compiled | Whole program is translated to machine code *before* running | C, C++, Go, Rust |
| Interpreted | Program is translated and run *line by line* | Python, JavaScript, Ruby |

## What is Python?
**Python** is a high-level, interpreted, general-purpose programming language created by **Guido van Rossum** and first released in **1991**. It is named after the comedy group *Monty Python*, not the snake.

Python's design philosophy stresses **readability** and simplicity. The same task that takes many lines in other languages can often be written in a few lines of Python.

```python
print("Hello, World!")   # one line is enough to display text
```

## Why learn Python?
- **Easy to read and learn:** the syntax looks close to plain English.
- **Versatile:** used in web development, data science, machine learning / AI, automation, cybersecurity, game development and more.
- **Huge community and libraries:** tools like NumPy, Pandas, Django, Flask and TensorFlow are ready to use.
- **Cross-platform:** the same code runs on Windows, macOS and Linux.
- **Dynamically typed:** you do not have to declare a variable's type; Python works it out for you.
- **In demand:** one of the most popular languages in the world.

## Key features at a glance
| Feature | Meaning |
|---------|---------|
| Interpreted | Code runs line by line, so errors are found quickly |
| Dynamically typed | `x = 5` then `x = "hello"` is allowed |
| Indentation-based | Spaces/tabs define code blocks instead of `{ }` |
| Case-sensitive | `Name` and `name` are different identifiers |
| Open source | Free to download, use and modify |

# Print & Escape Sequences


## Description
The `print()` function is the most basic way to **display output** on the screen. Whatever you place inside the brackets is converted to text and shown to the user. It is usually the first function every Python learner uses and the main tool for checking what your program is doing.

**Key points**
- Text (a *string*) must be wrapped in quotes: single `' '`, double `" "`, or triple `""" """`.
- `print()` can take several values separated by commas and adds a space between them.
- By default `print()` ends with a new line. You can change this with the `end` parameter, and change the separator with `sep`.
- Triple quotes let you write text over **multiple lines**.
- f-strings (`f"..."`) let you insert variables directly into text using `{ }`.

### Escape sequences
An **escape sequence** is a special combination that starts with a backslash `\` and has a special meaning inside a string.

| Escape | Meaning | Example output |
|--------|---------|----------------|
| `\n` | New line | moves text to the next line |
| `\t` | Tab (horizontal space) | aligns text in columns |
| `\\` | A literal backslash | `\` |
| `\'` | A single quote | `'` |
| `\"` | A double quote | `"` |
| `\xhh` | Character from a hex code | `\x41` gives `A` |
| `\uhhhh` | Unicode character | `\u2665` gives a heart |

## Examples

**1. Basic print**
```python
print("Welcome to Python")
```
Output:


**4. Triple quotes (multi-line text)**
```python

```
Output:
```

```

**5. f-string**


> **Tip:** If a Windows file path shows an error or odd output, the backslashes are being read as escape sequences. Double them (`\\`) or use a raw string such as `r"C:\new_folder"`.

# Variables


## Description
A **variable** is a named container in memory that stores a value. Think of it as a label stuck on a box: you can put something inside, look at it later, or replace it with something else. In Python a variable is created the moment you assign a value to it using the `=` operator, and you do **not** need to declare its type first.

**Key points**
- Syntax: `variable_name = value`
- Python is **dynamically typed**: the same variable can hold an `int`, then a `str`, then a `list`.
- Use `type(x)` to check the type and `id(x)` to see the memory address of the value.
- `del x` removes a variable. Using it afterwards raises a `NameError`.
- Constants (values that should not change) are written in UPPERCASE by convention, e.g. `PI = 3.14159`.

### Rules for naming variables
- Must start with a **letter or underscore** (`_`), never a digit.
- Can contain only letters, digits and underscores (no spaces, no `-`).
- Are **case-sensitive**: `Total` and `total` are different variables.
- Cannot be a Python **keyword** such as `class`, `def`, `if`, `del`, `for`.

| Name | Valid? | Reason |
|------|--------|--------|
| `my_name` | Yes | letters and underscore |
| `_count` | Yes | may start with underscore |
| `total_1` | Yes | digits allowed after the first character |
| `2name` | No | starts with a digit |
| `my-var` | No | hyphen is not allowed |
| `class` | No | reserved keyword |

### Naming style
Python programmers use **snake_case** (lowercase words joined by underscores), e.g. `student_age`, and choose meaningful names.

# Input-Based Programs

## Description
Programs become useful when they can **interact with the user**. The built-in `input()` function pauses the program, shows an optional message (prompt), waits for the user to type something and press Enter, and then returns what was typed.

**Key points**
- `input()` **always returns a string (`str`)**, even if the user types a number.
- To do maths you must **convert** (type-cast) the value using `int()`, `float()`, etc.
- `str.split()` breaks a line of text into a list of words; `map()` applies a function (like `int`) to each item.
- Combine `input()` with f-strings to produce friendly, formatted output.
- Basic arithmetic operators: `+` add, `-` subtract, `*` multiply, `/` divide (float result), `//` floor division, `%` remainder, `**` power.

### Typical flow of an input-based program
1. **Input:** read values from the user.
2. **Process:** convert types and perform calculations.
3. **Output:** display the result using `print()`.

## Examples

**1. Reading text**
```python
name = input("Enter your name: ")
print(f"Hello, {name}!")
```
Sample run:
```
Enter your name: Ravi
Hello, Ravi!
```

**2. Reading numbers (type conversion is needed)**
```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print("Sum =", a + b)
```
Sample run:
```
Enter first number: 7
Enter second number: 3
Sum = 10
```
> Without `int()`, typing `7` and `3` would give `73` because the values would be joined as text.

**3. Reading many values on one line**
```python
a, b, c = map(int, input("Enter three numbers: ").split())
print("Total =", a + b + c)
```
Sample run:
```
Enter three numbers: 4 5 6
Total = 15
```

**4. Splitting a full name**
```python
full = input("Full name: ")
parts = full.split()
print("First:", parts[0], "| Last:", parts[-1])
```
Sample run:
```
Full name: Ada Lovelace
First: Ada | Last: Lovelace
```

**5. Formatting numbers to 2 decimal places**
```python
price = float(input("Price: "))
print(f"Price with tax: {price * 1.05:.2f}")
```

> **Note:** If the user types text where a number is expected (for example `abc` for `int(input())`), Python raises a `ValueError`.
