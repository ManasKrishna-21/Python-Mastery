# Number Functions

## Description
A **number function** takes one or more numbers, performs a calculation and **returns** a result. Python gives you two families of them.

### 1. Built-in functions (always available)
| Function | What it does | Example | Result |
|----------|--------------|---------|--------|
| `abs(x)` | Absolute value (distance from 0) | `abs(-7.5)` | `7.5` |
| `round(x)` | Nearest whole number (returns an `int`) | `round(7.4)` | `7` |
| `round(x, n)` | Round to `n` decimal places (`n` can be negative) | `round(3.14159, 2)` | `3.14` |
| `pow(x, y)` | x to the power y (same as `x ** y`) | `pow(2, 10)` | `1024` |
| `pow(x, y, m)` | `(x ** y) % m`, computed efficiently | `pow(2, 5, 7)` | `4` |
| `divmod(a, b)` | Quotient and remainder together | `divmod(17, 5)` | `(3, 2)` |
| `min()` / `max()` | Smallest / largest value | `max(4, 9, 2)` | `9` |
| `sum(iterable)` | Total of all items | `sum([4, 9, 2])` | `15` |

### Rounding: four different ways
For a decimal number, each function gives a different answer. Compare them for `3.7` and `-3.7`:

| Function | `3.7` | `-3.7` | Rule |
|----------|-------|--------|------|
| `round()` | `4` | `-4` | Nearest whole number |
| `math.ceil()` | `4` | `-3` | Always **up** (towards +infinity) |
| `math.floor()` | `3` | `-4` | Always **down** (towards -infinity) |
| `math.trunc()` / `int()` | `3` | `-3` | Cut off the decimals (towards **zero**) |

### Important ideas
- **Round half to even ("banker's rounding"):** when a number is exactly halfway, `round()` picks the **even** neighbour. So `round(2.5)` is `2`, `round(3.5)` is `4`, `round(4.5)` is `4`. This avoids always rounding upwards.
- **Floats are not exact:** `0.1` cannot be stored exactly in binary, so `0.1 + 0.2` is `0.30000000000000004`. Because of this, `round(2.675, 2)` gives `2.67`, not `2.68`. Use `math.isclose()` instead of `==` to compare floats.
- **Trigonometric functions use radians,** not degrees: `180 degrees = pi radians`. Convert with `math.radians()` and `math.degrees()`.
- **`math.sqrt(-1)` raises a `ValueError`** ("math domain error") because the result is not a real number.
- **`math.sqrt` returns a float,** which loses accuracy for very large integers; `math.isqrt` is exact.
- `comb(n, r)` counts selections where **order does not matter**; `perm(n, r)` counts arrangements where **order matters**. So `perm = comb x r!`.
- **`divmod(a, b)`** is shorthand for `(a // b, a % b)` and is ideal for unit conversions such as minutes to hours and minutes.
- **Digits of an integer:** a positive integer `n` has `floor(log10(n)) + 1` digits.

### Problem 1: `abs()`, `round()` and `pow()` on Integers and Floats
**Problem statement:** Demonstrate `abs()`, `round()` and `pow()` on a mix of integers and floats, including negative values.

**Definition:** `abs(x)` returns the **absolute value**: the distance of `x` from zero, so it is never negative. `round(x)` returns the **nearest whole number** as an `int`. `pow(x, y)` returns `x` raised to the power `y` and is the same as `x ** y`.

**Examples:**
- `abs(-21)` → `21`, `abs(-8.45)` → `8.45`
- `round(9.75)` → `10`, `round(-8.45)` → `-8`
- `pow(30, 2)` → `900` (squared), `pow(-21, 3)` → `-9261` (cubed, the sign is kept)

**Approach:**
1. Store a mix of positive/negative integers and floats in variables.
2. Print `abs()` of each, then `round()` of each.
3. Print `pow(x, 2)` and `pow(x, 3)` for each.

**Sample input → output:**
- `(no input)` → `30 21 9.75 8.45 / 30 -21 10 -8 / 900 441 95.0625 71.40249999999999 / 27000 -9261 926.859375 -603.3511249999998`

> **Note on the solution below:** powers of decimals such as `-8.45` show tiny float errors (`71.40249999999999` instead of `71.4025`). This is normal for floating-point numbers.

````python
# 1.Demonstrate abs(), round(), and pow() on a mix of integers and floats.
# A mix of integers (30, -21) and floats (9.75, -8.45) to see how each function treats both
num1 = 30
num2 = -21
num3 = 9.75
num4 = -8.45
# abs() returns the distance from zero, so the minus sign is dropped: 30 21 9.75 8.45
print(f"{abs(num1)} {abs(num2)} {abs(num3)} {abs(num4)}")
# round() with one argument gives the nearest whole number as an int (9.75 -> 10, -8.45 -> -8)
print(f"{round(num1)} {round(num2)} {round(num3)} {round(num4)}")
# pow(x, 2) is x squared (same as x ** 2); a square is never negative
# Decimals can show tiny float errors, e.g. 71.40249999999999 instead of 71.4025
print(f"{pow(num1,2)} {pow(num2,2)} {pow(num3,2)} {pow(num4,2)}")
# pow(x, 3) is x cubed; the sign of the original number is kept (negative numbers give negative cubes)
print(f"{pow(num1,3)} {pow(num2,3)} {pow(num3,3)} {pow(num4,3)}")
````

### Problem 2: `round()` with `ndigits` and the Tie Rule
**Problem statement:** Use `round()` with a number of digits (including negative values). Compare `round(2.5)` with `round(3.5)` and explain the result.

**Definition:** `round(x, n)` rounds to `n` digits after the decimal point. A **negative** `n` rounds to the left of the decimal point: `-1` means the nearest 10, `-2` the nearest 100, and so on. When a number is exactly halfway between two choices, Python uses **round half to even** (also called *banker's rounding*): the tie goes to the **even** neighbour.

**Examples:**
- `round(1234.14159, 2)` → `1234.14`
- `round(1234.14159, -2)` → `1200.0` (nearest hundred)
- `round(2.5)` → `2` and `round(3.5)` → `4` (both go to the even number)
- `round(4.5)` → `4`, `round(5.5)` → `6`

**Approach:**
1. Round the same number with `ndigits` = 0, 2, 3, -1, -2, -3 and observe each result.
2. Round 2.5, 3.5, 4.5, 5.5 and look at which neighbour is chosen.
3. Explain: ties are rounded to the even neighbour, which avoids always rounding up.

**Sample input → output:**
- `(no input)` → `1234 / 1234.14 / 1234.142 / 1230.0 / 1200.0 / 1000.0 / 2 / 4 / 4 / 6`

````python
# 2. Use round() with ndigits (including negative values). Observe round(2.5) versus round(3.5) and explain it.
# The second argument (ndigits) sets where rounding happens:
# positive = digits after the decimal point, negative = digits to the left of it
# No ndigits -> nearest whole number: 1234
print(round(1234.14159))
# 2 decimal places -> 1234.14
print(round(1234.14159,2))
# 3 decimal places -> 1234.142
print(round(1234.14159,3))
# -1 rounds to the nearest 10 -> 1230.0 (a float, because the input is a float)
print(round(1234.14159,-1))
# -2 rounds to the nearest 100 -> 1200.0
print(round(1234.14159,-2))
# -3 rounds to the nearest 1000 -> 1000.0
print(round(1234.14159,-3))
# If a number is exactly halfway between two choices,
# Python breaks the tie by choosing the even digit at the target precision:
# Results below: 2, 4, 4, 6. Every tie goes to the nearest EVEN number ('banker's rounding'),
# which avoids always rounding half-values upwards
print(round(2.5))
print(round(3.5))
print(round(4.5))
print(round(5.5))
````

### Problem 3: `divmod()`: Minutes to Hours and Minutes
**Problem statement:** Use `divmod()` to convert a number of minutes (for example 500) into hours and minutes.

**Definition:** `divmod(a, b)` returns a pair `(quotient, remainder)`, which is the same as `(a // b, a % b)` calculated in one step. With `b = 60`, the quotient is the number of **whole hours** and the remainder is the **leftover minutes**.

**Examples:**
- `divmod(500, 60)` → `(8, 20)`, so 500 minutes = 8 hours 20 minutes
- `divmod(45, 60)` → `(0, 45)`
- `divmod(120, 60)` → `(2, 0)`

**Approach:**
1. Read the number of minutes and convert it with `int()`.
2. Call `divmod(minutes, 60)` and unpack the pair into `hours` and `mins`.
3. Print both values.

**Sample input → output:**
- `500` → `8 hours 20 minutes`

````python
# 3. Use divmod() to convert 500 minutes into hours and minutes.
# input() returns text, so int() converts it into a number
minutes = int(input("Enter the no of minutes : "))
# divmod(a, b) returns (a // b, a % b) in one step: (whole hours, leftover minutes)
# For 500 minutes: divmod(500, 60) = (8, 20)
hours, mins = divmod(minutes,60)
print(f"{hours} hours {mins} minutes")
````

### Problem 5: Solving a Quadratic Equation with `math.sqrt()`
**Problem statement:** Read the coefficients `a`, `b` and `c` of a quadratic equation and print its real roots, including the case with no real roots.

**Definition:** A **quadratic equation** has the form `ax² + bx + c = 0` (with `a ≠ 0`). Its **discriminant** is `D = b² - 4ac`. If `D > 0` there are **two real roots**, if `D = 0` there is **one repeated root**, and if `D < 0` there are **no real roots**. The roots are `x = (-b ± √D) / (2a)`.

**Examples:**
- `x² - 3x + 2 = 0` (a=1, b=-3, c=2): D = 1 → roots `2.0` and `1.0`
- `x² + 2x + 1 = 0`: D = 0 → one root `-1.0`
- `x² + 1 = 0`: D = -4 → no real roots

**Approach:**
1. Calculate `discriminant = b**2 - 4*a*c`.
2. If it is negative, print "No Real Roots".
3. If it is zero, print the single root `-b / (2a)`.
4. Otherwise compute both roots with `math.sqrt(discriminant)`.

**Sample input → output:**
- `1, -3, 2` → `Root 1 : 2.0 / Root 2 :1.0`
- `1, 2, 1` → `Roots : -1.0`
- `1, 0, 1` → `No Real Roots`

> **Note on the solution below:** entering `a = 0` is not a quadratic equation and causes a `ZeroDivisionError`. Also, `math.sqrt()` is only called when the discriminant is positive, which avoids a `ValueError`.

````python
# 5. Use math.sqrt() to solve a quadratic equation, including the no-real-roots case.
# math.sqrt() lives in the math module, so it has to be imported first
import math
# Quadratic equation: a*x**2 + b*x + c = 0 (a must not be 0)
a = int(input("Enter first number of equation : "))
b = int(input("Enter second number of equation : "))
c = int(input("Enter third number of equation : "))
# Discriminant D = b**2 - 4ac decides the number of real roots: D < 0 none, D = 0 one repeated root, D > 0 two roots
discriminant = b**2 - 4*a*c
if discriminant < 0:
  # The square root of a negative number is not real (math.sqrt would raise a ValueError)
  print("No Real Roots")
elif discriminant == 0:
  # D = 0 gives one repeated root: x = -b / (2a)
  print(f"Roots : {(-b/(2*a))}")
else:
  # D > 0 gives two roots from the quadratic formula: x = (-b +/- sqrt(D)) / (2a)
  root1 = (-b + math.sqrt(discriminant)) / (2*a)
  root2 = (-b - math.sqrt(discriminant)) / (2*a)
  print(f"Root 1 : {root1}\nRoot 2 :{root2}")
````

### Problem 6: Buses Needed with `math.ceil()` and `math.floor()`
**Problem statement:** A bus carries 40 people. Find how many buses are needed for `n` people, and how many of them are completely full.

**Definition:** `math.ceil(x)` rounds **up** to the next whole number; `math.floor(x)` rounds **down**. For a division like `n / 40`, `ceil` gives the buses **needed** (a partly filled bus still counts) and `floor` gives the buses that are **completely full**.

**Examples:**
- 135 people: 135 / 40 = 3.375 → `ceil` = **4** buses needed, `floor` = **3** full buses
- 120 people: 120 / 40 = 3.0 → 3 buses needed and 3 full
- 1 person: 1 / 40 = 0.025 → 1 bus needed, 0 full

**Approach:**
1. Read `n` and set `cap = 40`.
2. Total buses = `math.ceil(n / cap)`.
3. Full buses = `math.floor(n / cap)`.

**Sample input → output:**
- `135` → `Total Buses : 4 / Full Buses : 3`

````python
# 6. Use math.ceil() and math.floor() to work out how many buses (capacity 40) are needed for n people.
# ceil() rounds UP and floor() rounds DOWN to a whole number
import math
n = int(input("Enter the number of people : "))
# Each bus carries 40 people
cap = 40
# ceil(n / cap) rounds up, because a partly filled bus is still a bus (135 people: 3.375 -> 4 buses)
print(f"Total Buses : {math.ceil(n/cap)}")
# floor(n / cap) counts only the completely full buses (135 people: 3 full buses)
print(f"Full Buses : {math.floor(n/cap)}")
````

### Problem 7: Factorial, Combinations and Permutations
**Problem statement:** Compute nCr and nPr using `math.comb()` and `math.perm()`, and verify them with `math.factorial()`.

**Definition:** The **factorial** `n!` is `1 × 2 × ... × n`. A **combination** `nCr = n! / (r! (n-r)!)` is the number of ways to **choose** `r` items from `n` when **order does not matter**. A **permutation** `nPr = n! / (n-r)!` is the number of ways to **arrange** `r` items out of `n` when **order matters**.

**Examples:**
- `6C2` = 15 (choose 2 players from 6)
- `6P2` = 30 (arrange 2 of 6 people in a line: 6 × 5)
- `5! = 120`
- `nPr = nCr × r!`, so `30 = 15 × 2!`

**Approach:**
1. Call `math.comb(n, r)` and `math.perm(n, r)`.
2. Calculate the same values manually with `math.factorial()`.
3. Both methods must give the same answers.

**Sample input → output:**
- `n = 6, r = 2` → `Permutation : 30 / Combination : 15 (printed twice, once for each method)`

> **Note on the solution below:** this cell uses `math` without importing it, because it was imported in an earlier cell. If you run this cell alone, run `import math` first.

````python
# 7. Use math.factorial(), math.comb(), and math.perm() to compute nCr and nPr.
# Note: math was imported in an earlier cell. If you run this cell on its own, run 'import math' first
# n = total items, r = items chosen
n,r = 6,2
# comb(n, r) = n! / (r! * (n-r)!): ways to CHOOSE r items from n when order does NOT matter
ncr = math.comb(n,r)
# perm(n, r) = n! / (n-r)!: ways to ARRANGE r items out of n when order DOES matter
npr = math.perm(n,r)
# Expected for n = 6, r = 2: Permutation 30, Combination 15
print(f"Permutation : {npr}\nCombination : {ncr}")

# The same results from the factorial formulas ('//' keeps the result an integer); they must match the lines above
ncr_manual = math.factorial(n) // (math.factorial(r) * math.factorial(n - r))
npr_manual = math.factorial(n) // math.factorial(n - r)
print(f"Permutation : {npr_manual}\nCombination : {ncr_manual}")
````

### Problem 8: GCD and LCM with the `math` Module
**Problem statement:** Use `math.gcd()` and `math.lcm()` and compare the results with your own functions from the earlier number problems.

**Definition:** The **GCD** (greatest common divisor) is the largest number that divides both numbers exactly. The **LCM** (least common multiple) is the smallest positive number that both numbers divide exactly. They are linked by `GCD × LCM = a × b`.

**Examples:**
- GCD(12, 18) = **6**
- LCM(12, 18) = **36** (check: 12 × 18 / 6 = 36)
- GCD(7, 13) = 1 (co-prime numbers)

**Approach:**
1. Call `math.gcd(12, 18)` and `math.lcm(12, 18)`.
2. Compare with your own versions written with loops (Day 5, Problems 7 and 8); the answers must be identical.

### Problem 9: Triangle Height from an Angle and a Base
**Problem statement:** Use `math.pi`, `sin`, `cos`, `radians` and `degrees` to find the height of a right-angled triangle from a base and an angle, then its area.

**Definition:** In a **right-angled triangle**, the **tangent** of an angle is `opposite / adjacent`, and `tan(θ) = sin(θ) / cos(θ)`. So with the angle `θ` at the base: `height = base × tan(θ)`. The area is `½ × base × height`. Python's trigonometric functions work in **radians** (`180° = π radians`); `math.radians()` and `math.degrees()` convert between the two.

**Examples:**
- Base = 10, angle = 30° (π/6 radians): height = 10 × tan(30°) ≈ **5.7735**
- Area = ½ × 10 × 5.7735 ≈ **28.8675**
- `math.degrees(math.pi / 6)` ≈ `30.0`

**Approach:**
1. Convert the angle to radians (here `math.pi / 6`).
2. Convert to degrees and back to radians to practise both functions.
3. Compute `height = base * sin / cos`.
4. Compute and print the area `0.5 * base * height`.

**Sample input → output:**
- `(no input)` → `Area of Triangle : 28.86751345948128`

> **Note on the solution below:** the question asks for the height, which is stored in the variable `height` (about 5.77), but the final line prints the **area**. Print `height` as well to see it.

````python
# 9. Use math.pi, sin, cos, radians, and degrees to find a triangle’s height from an angle and a base.
import math

# Right-angled triangle: the base is the side next to the angle and the height is the side opposite to it
base = 10.0
# pi radians = 180 degrees, so pi / 6 radians = 30 degrees (trig functions work in radians)
angle_rad = math.pi / 6
# degrees() converts radians to degrees: about 30.0
angle_deg = math.degrees(angle_rad)
# radians() converts back, so the two conversions cancel each other out
rad_again = math.radians(angle_deg)
# tan(angle) = sin(angle) / cos(angle) = height / base, so height = base x tan(angle), about 5.77
height = base * (math.sin(rad_again) / math.cos(rad_again))
# Area = 0.5 x base x height. Note: the question asks for the height, which is stored in 'height' (print it to see it)
print(f"Area of Triangle : {0.5*base*height}")
````

**Problem statement:** Use `math.isqrt()` to check whether a number is a perfect square.

**Definition:** A **perfect square** is an integer that is the square of another integer (such as 144 = 12²). `math.isqrt(n)` returns the **integer square root**: the largest integer `r` with `r × r ≤ n`. If `r * r == n`, then `n` is a perfect square. Unlike `math.sqrt()`, `isqrt` is **exact even for huge numbers**.

### Problem 13: Float Accuracy: `math.fsum()` and `math.isclose()`
**Problem statement:** Compare `sum([0.1]*10)` with `math.fsum()`, and use `math.isclose()` to compare floats safely.

**Definition:** Computers store decimals in binary, so numbers like `0.1` are stored with a tiny error and repeated additions can add up to something slightly different from the exact answer. `math.fsum()` adds floats with extra precision. `math.isclose(a, b)` returns `True` if `a` and `b` are **close enough** (within a relative tolerance of `1e-09` by default). Comparing with `0` needs an **absolute tolerance** (`abs_tol`).

**Examples:**
- Adding `0.1` ten times in a plain loop gives `0.9999999999999999`, not `1.0`
- `math.fsum([0.1]*10)` gives exactly `1.0`
- `math.isclose(0.1 + 0.2, 0.3)` → `True` (even though `0.1 + 0.2 == 0.3` is `False`)
- `math.isclose(0.0, 1e-12)` → `False`, but with `abs_tol=1e-9` → `True`

**Approach:**
1. Sum ten copies of `0.1` with `sum()`, `math.fsum()` and a plain loop.
2. Compare each total with `1.0` using `==`.
3. Compare with `math.isclose()`, with and without `abs_tol`.

**Sample input → output:**
- `(no input)` → `sum = 1.0, fsum = 1.0, ..., isclose(0.0, 1e-12) False, with abs_tol True, naive loop: 0.9999999999999999`

> **Note on the solution below:** from Python **3.12**, `sum()` also uses compensated summation, so `sum([0.1]*10)` prints `1.0`. In older versions it prints `0.9999999999999999`. The plain `+=` loop at the end shows the error on every version.

````python
# 13. Compare sum([0.1]*10) with math.fsum(), and use math.isclose() to compare floats.
import math

# Floats cannot store 0.1 exactly, so repeated additions can build up a tiny error (see the plain loop at the end)
# Note: Python 3.12+ sum() uses compensated summation and gives 1.0; older versions give 0.9999999999999999
print("sum([0.1]*10)     =", repr(sum([0.1]*10)))
# fsum() keeps track of the lost digits while adding, so it returns the correct total 1.0 on every Python version
print("math.fsum([0.1]*10)=", repr(math.fsum([0.1]*10)))
# Comparing floats with == is risky: this is True on Python 3.12+ but False on older versions
print("sum == 1.0 :", sum([0.1]*10) == 1.0)
# True on every version
print("fsum == 1.0:", math.fsum([0.1]*10) == 1.0)
# isclose() treats two floats as equal when they are within a small relative tolerance (1e-09 by default)
# so it is the safe way to compare floats
print("isclose(sum, 1.0):", math.isclose(sum([0.1]*10), 1.0))
# 0.1 + 0.2 is really 0.30000000000000004, yet isclose() accepts it as 0.3: True
print("isclose(0.1+0.2, 0.3) :", math.isclose(0.1+0.2, 0.3))
# A relative tolerance never works when comparing with 0, so 1e-12 is not 'close' to 0 by default: False
print("isclose(0.0, 1e-12) abs_tol=0:", math.isclose(0.0, 1e-12))
# An absolute tolerance fixes it: differences up to 1e-9 are accepted: True
print("isclose(0.0, 1e-12, abs_tol=1e-9):", math.isclose(0.0, 1e-12, abs_tol=1e-9))

# A plain loop adding 0.1 ten times shows the classic float error: 0.9999999999999999
t = 0.0
for _ in range(10):
    t += 0.1
print("naive loop:", repr(t))
````

### Problem 14: Decimal to Any Base (2 to 16) and Back
**Problem statement:** Convert a number from decimal to any base from 2 to 16 and convert it back to decimal.

**Definition:** A **base** (radix) is the number of digit symbols a number system uses: base 2 uses `0-1`, base 8 uses `0-7`, base 16 uses `0-9` and `A-F`. To convert a decimal number to base `b`: **divide by `b` repeatedly**; the remainders, read from last to first, are the digits. Converting back is done with `int(text, base)`.

**Examples:**
- `255` in base 2 → `11111111`
- `255` in base 16 → `FF`
- `1000` in base 7 → `2626` (2×343 + 6×49 + 2×7 + 6)
- `-42` in base 5 → `-132` (the minus sign is added in front)

**Approach:**
1. Take the absolute value of the number.
2. While it is not 0: `m, r = divmod(m, b)` and save the digit `DIGITS[r]`.
3. Reverse the saved digits and add `-` for negative numbers (use `0` if there are no digits).
4. Verify with `int(s, b)`.

**Sample input → output:**
- `(no input)` → `255 -> base 2: 11111111, 255 -> base 16: FF, 1000 -> base 7: 2626, -42 -> base 5: -132 (all converted back correctly)`

````python
# 14. Convert a number from decimal to any base (2 to 16) and back.
# Digit symbols for bases up to 16: 0-9, then A-F for the values 10-15
DIGITS = "0123456789ABCDEF"

# Test cases as (number, base) pairs, including a negative number
for n, b in [(255, 2), (255, 16), (1000, 7), (-42, 5)]:
    # Convert the absolute value; the minus sign is added back at the end
    m = abs(n)
    # The remainders are collected here, last digit first
    digits = []
    # Repeat until m becomes 0 (0 counts as False)
    while m:
        # Divide by the base: the remainder is the next digit from the right and the quotient is carried on
        m, r = divmod(m, b)
        digits.append(DIGITS[r])
    # The remainders come out in reverse order, so reverse them; if there are none (n = 0) use '0'
    s = ("-" if n < 0 else "") + ("".join(reversed(digits)) or "0")
    # int(text, base) converts back to decimal, proving that the conversion was correct
    back = int(s, b)
    # {b:>2} and {s:>10} right-align the values so that the output lines up in columns
    print(f"{n} -> base {b:>2}: {s:>10} -> back: {back}  ok={back == n}")
````

### Problem 15: Distance Between Points and Heron's Formula
**Problem statement:** Find the distance between two points with `math.dist()` or `math.hypot()`, and the area of a triangle using Heron's formula.

**Definition:** The **distance** between points `(x1, y1)` and `(x2, y2)` is `√((x2-x1)² + (y2-y1)²)` (Pythagoras). `math.dist(p, q)` does this directly, and `math.hypot(dx, dy)` gives the length of the hypotenuse for the differences `dx` and `dy`. **Heron's formula** gives the area of any triangle from its three sides `a, b, c`: with `s = (a + b + c) / 2`, **area = √(s(s-a)(s-b)(s-c))**.

**Examples:**
- Distance from `(1, 2)` to `(4, 6)` = √(3² + 4²) = **5.0**
- Triangle with corners `(0,0)`, `(3,0)`, `(0,4)`: sides 3, 5, 4 → s = 6 → area = √(6×3×1×2) = **6.0**

**Approach:**
1. Create the two points as tuples and call `math.dist()` and `math.hypot()`.
2. Get the three side lengths of the triangle with `math.dist()`.
3. Calculate `s` and then the area with Heron's formula.

**Sample input → output:**
- `(no input)` → `math.dist: 5.0 / math.hypot: 5.0 / sides: 3.0 5.0 4.0 / area: 6.0`

````python
# 15. Find the distance between two points with math.dist() or math.hypot(), and the area of a triangle using Heron’s formula.
import math

# Two points written as (x, y) tuples
p, q = (1, 2), (4, 6)
# dist(p, q) = sqrt((x2 - x1)**2 + (y2 - y1)**2) = sqrt(3**2 + 4**2) = 5.0
print("math.dist:", math.dist(p, q))
# hypot(dx, dy) gives the same result from the differences along each axis (the hypotenuse length)
print("math.hypot:", math.hypot(q[0]-p[0], q[1]-p[1]))

# Triangle with corners (0,0), (3,0) and (0,4): find the three side lengths with dist()
a = math.dist((0,0),(3,0))
b = math.dist((3,0),(0,4))
c = math.dist((0,4),(0,0))
# Heron's formula: s is the semi-perimeter (half of the perimeter)
s = (a + b + c) / 2
# area = sqrt(s(s-a)(s-b)(s-c)); for this 3-4-5 right triangle the area is 6.0 (= 0.5 x 3 x 4)
area = math.sqrt(s * (s-a) * (s-b) * (s-c))
print("sides:", a, b, c, "\narea:", area)
````
