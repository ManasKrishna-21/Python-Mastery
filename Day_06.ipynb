# Strings


## Description
A **string** is an ordered sequence of characters (letters, digits, spaces and symbols) enclosed in single `' '`, double `" "` or triple `""" """` quotes.

```python
name = "Python"
```

### Key properties
- **Ordered:** every character has a position called an **index**, starting from `0`.
- **Immutable:** a string cannot be changed after creation. Methods such as `upper()` return a **new** string and leave the original unchanged.
- **Iterable:** you can loop over a string character by character with `for ch in text:`.
- **Case-sensitive:** `"A"` and `"a"` are different characters.

### Indexing and slicing
For `s = "Python"`:

| Character | P | y | t | h | o | n |
|-----------|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 | 5 |
| Negative index | -6 | -5 | -4 | -3 | -2 | -1 |

| Expression | Meaning | Result |
|------------|---------|--------|
| `s[0]` | first character | `'P'` |
| `s[-1]` | last character | `'n'` |
| `s[1:4]` | from index 1 up to (not including) 4 | `'yth'` |
| `s[:3]` | first three characters | `'Pyt'` |
| `s[::2]` | every second character | `'Pto'` |
| `s[::-1]` | whole string reversed | `'nohtyP'` |

### Commonly used string methods
| Group | Methods | Purpose |
|-------|---------|---------|
| **Case** | `upper()`, `lower()`, `title()`, `swapcase()`, `capitalize()` | Change the letter case |
| **Cleaning** | `strip()`, `lstrip()`, `rstrip()` | Remove spaces from both / left / right ends |
| **Split and join** | `split()`, `"sep".join(list)` | Break text into words, or glue words together |
| **Search** | `count()`, `find()`, `startswith()`, `endswith()`, `in` | Look for text inside a string |
| **Replace** | `replace(old, new)` | Swap one piece of text for another |
| **Character tests** | `isalpha()`, `isdigit()`, `isspace()`, `isupper()`, `islower()` | Return `True` or `False` for a character |
| **Length** | `len(s)` | Number of characters |

### Useful helper functions
| Function | Use |
|----------|-----|
| `ord(ch)` / `chr(code)` | Convert between a character and its numeric code (needed for ciphers) |
| `sorted(s)` | Returns the characters as a sorted list (used for anagrams) |
| `set(s)` | Keeps only unique characters (but loses the order) |
| `"".join(list)` | Turns a list of characters back into a string |

### Common patterns in string problems
- **Build a new string:** start with `result = ""` and add characters one by one (`result += ch`).
- **Counter:** start with `count = 0` and add 1 when a character matches.
- **Flag:** assume `True`, set to `False` when a mismatch is found (pangram, palindrome).
- **Compare with previous:** keep track of the previous character (compression).
- **Case-insensitive matching:** convert everything with `lower()` first.

### Common mistakes
- Trying to modify a string in place (`s[0] = "J"` raises a `TypeError`).
- Forgetting that `input()` always returns a string, even for digits.
- Off-by-one errors with slicing: the end index is **not** included.
- Forgetting that spaces and punctuation are also characters.

### Problem 1: String Length and Case Methods
**Problem statement:** For the string `"aLlen iverson"`, print its length and its upper, lower, title and swapcase versions.

**Definition:** The **length** of a string is the number of characters in it, including spaces. The case methods return a **new** string: `upper()` (all capitals), `lower()` (all small letters), `title()` (first letter of every word capital, the rest small) and `swapcase()` (capitals become small and small become capitals).

**Examples:**
- `len("Hi there")` → `8` (the space counts)
- `"python".upper()` → `PYTHON`
- `"hELLO wORLD".title()` → `Hello World`
- `"PyThOn".swapcase()` → `pYtHoN`

**Approach:**
1. Store the text in a variable.
2. Print `len()` of the variable.
3. Call each case method and print the returned string. The original variable stays unchanged.

**Sample input → output:**
- `(no input)` → `13 / ALLEN IVERSON / allen iverson / Allen Iverson / AlLEN IVERSON`

````python
# 1. Print the length of a string and its upper, lower, title, and swapcase versions.
# Strings are immutable: every method below returns a NEW string and leaves name unchanged
name = "aLlen iverson"
# len() counts every character, including the space: 13
print(len(name))
# upper() makes all letters capital: ALLEN IVERSON
print(name.upper())
# lower() makes all letters small: allen iverson
print(name.lower())
# title() capitalises the first letter of every word: Allen Iverson
print(name.title())
# swapcase() flips the case of each letter: AlLEN IVERSON
print(name.swapcase())
````

### Problem 2: Reverse a String
**Problem statement:** Reverse a string using slicing, then again using a loop.

**Definition:** The **reverse** of a string has its characters in the opposite order. There are two common ways: a **slice with step -1** (`s[::-1]`), or a **loop** that adds each new character to the *front* of a result string.

**Examples:**
- `"Allen"` → `"nellA"`
- `"abc"` → `"cba"`
- `"a"` → `"a"` (one character is its own reverse)

**Approach:**
1. Slicing: `name[::-1]`.
2. Loop: start with `rev = ""`; for every character do `rev = char + rev`.
3. Both methods must give the same result.

**Sample input → output:**
- `(no input)` → `Slicing : nella, then: Reversed String : nella (the code also converts to lowercase)`

````python
# 2. Reverse a string using slicing, then again using a loop.
# Two ways to reverse: (1) slicing with step -1, (2) a loop that builds the result
name = "Allen"
# [::-1] means 'whole string, stepping backwards'; lower() only makes the output lowercase
print(f"Slicing : {name[::-1].lower()}")
# Method 2: start with an empty string and add each character to the FRONT of it
rev = ""
for char in name:
  # New character goes before the existing result: 'a' -> 'la' -> 'lla' -> ... -> 'nella'
  rev = char.lower() + rev
print(f"Reversed String : {rev}")
````

### Problem 3: Palindrome String
**Problem statement:** Check whether a string is a palindrome, ignoring case and spaces.

**Definition:** A **palindrome** is a word or phrase that reads the **same forwards and backwards**. Ignoring case means `M` and `m` are treated alike. Ignoring spaces means `"nurses run"` is read as `"nursesrun"`.

**Examples:**
- `madam`, `level`, `racecar` are palindromes
- `"Madam"` is a palindrome once the case is ignored
- `"nurses run"` is a palindrome once the space is ignored
- `hello` is not a palindrome (reverse is `olleh`)

**Approach:**
1. Convert the text to lowercase (and remove spaces).
2. Build its reverse.
3. If the text equals its reverse, it is a palindrome.

**Sample input → output:**
- `Madam` → `madam is Palindrome`
- `hello` → `hello is not Palindrome`

> **Note on the solution below:** this version converts to lowercase but does **not** remove spaces, so a phrase like `"nurses run"` is reported as not a palindrome. Add `.replace(" ", "")` after `.lower()` to ignore spaces too.

````python
# 3. Check whether a string is a palindrome (ignore case and spaces).
# lower() makes the check case-insensitive ('Madam' and 'madam' are treated the same)
# Spaces are NOT removed here; to ignore them as well add .replace(' ', '') after lower()
name = input("Enter your name : ").lower()
# Build the reverse of the text
rev = ""
for char in name:
  rev = char.lower() + rev
# Palindrome: the text reads the same forwards and backwards
if name == rev:
  print(f"{name} is Palindrome")
else:
  print(f"{name} is not Palindrome")
````

### Problem 4: Count Vowels, Consonants, Digits and Spaces
**Problem statement:** Read a string and count how many vowels, consonants, digits and spaces it contains.

**Definition:** **Vowels** are `a, e, i, o, u`. **Consonants** are the other letters. **Digits** are the characters `0` to `9`. **Spaces** are whitespace characters. Symbols (such as `!` or `@`) belong to none of the four groups.

**Examples:**
- `"Hello World 123"` → 3 vowels (e, o, o), 7 consonants, 3 digits, 2 spaces
- `"AEIOU"` → 5 vowels (when both cases are handled)
- `"a1 b2!"` → 1 vowel, 1 consonant, 2 digits, 1 space (the `!` is not counted)

**Approach:**
1. Create four counters starting at 0.
2. Loop through each character and test it: vowel, letter (consonant), digit, space.
3. Increase the matching counter and print all four.

**Sample input → output:**
- `Hello World 123` → `Vowel 3, Consonant 7, Digit 3, Space 2`

> **Note on the solution below:** only **lowercase** vowels are listed, so an uppercase vowel (such as `A`) is counted as a consonant. Use `char.lower() in ['a','e','i','o','u']` to count both cases.

````python
# 4. Count vowels, consonants, digits, and spaces in a string.
word = input("Enter your word : ")
# One counter for each category
vow_count = 0
con_count = 0
dig_count = 0
spa_count = 0
# Examine every character once
for char in word:
  # Only lowercase vowels are listed, so an uppercase vowel such as 'A' is counted as a consonant
  # (use char.lower() in ['a','e','i','o','u'] to count both cases)
  if char in ['a','e','i','o','u']:
    vow_count += 1
  # A letter that is not a vowel is a consonant
  elif char.isalpha():
    con_count += 1
  # A character from 0 to 9
  elif char.isdigit():
    dig_count += 1
  # A space (or tab / newline)
  elif char.isspace():
    spa_count += 1
print(f"Vowel Count : {vow_count}")
print(f"Consonant Count : {con_count}")
print(f"Digit Count : {dig_count}")
print(f"Space Count : {spa_count}")
````

### Problem 5: Frequency of a Character
**Problem statement:** Count how many times a given character appears in a string.

**Definition:** The **frequency** of a character is the number of times it occurs in the string. Counting is case-sensitive: `"A"` and `"a"` are different characters.

**Examples:**
- In `"banana"`, the frequency of `a` is **3**, of `n` is **2** and of `b` is **1**
- In `"Hello"`, the frequency of `l` is **2** and of `L` is **0**

**Approach:**
1. Read the word and the character to check.
2. Set `count = 0`.
3. Loop through the word and add 1 whenever a letter equals the character.
4. Print the count.

**Sample input → output:**
- `banana, a` → `The banana consists of a : 3`

> **Note on the solution below:** this solution counts **one chosen character**. To print the frequency of **every** character, use a dictionary: `freq[ch] = freq.get(ch, 0) + 1` (see Example 9 above).

````python
# 5. Print the frequency of each character in a string.
# Reads a word and ONE character, then counts how many times that character appears
# To get the frequency of EVERY character use a dictionary: freq[ch] = freq.get(ch, 0) + 1
word = input("Enter the word : ")
char = input("Enter character to be checked : ")
count = 0
# Compare each letter with the chosen character (the comparison is case-sensitive)
for letter in word:
  if letter == char:
    count += 1
print(f"The {word} consists of {char} : {count}")
````

### Problem 6: Remove Duplicate Characters (Preserving Order)
**Problem statement:** Remove repeated characters from a string while keeping the original order of first appearance.

**Definition:** Each character is kept only the **first time** it appears; later repeats are dropped. A `set` is great for remembering what has been seen, but a set has no order, so a **list** is used to keep the order.

**Examples:**
- `"programming"` → `"progamin"`
- `"aabbcc"` → `"abc"`
- `"abc"` → `"abc"` (already unique)

**Approach:**
1. Create an empty `set` (seen characters) and an empty `list` (result).
2. For each character: if it is not in the set, add it to the set and the list.
3. Join the list into a string.

**Sample input → output:**
- `programming` → `Only Unique Characters : progamin`

````python
# 6. Remove duplicate characters from a string while preserving order.
word = input("Enter the word : ")
# A set remembers which characters were already seen (fast membership test)
unique = set()
# A list keeps the unique characters in their original order (a set does not keep order)
result = []
for char in word:
  # First time this character appears -> keep it
  if char not in unique:
    unique.add(char)
    result.append(char)
# join() glues the list of characters into one string
# Note: using the same double quotes inside an f-string needs Python 3.12 or newer
print(f"Only Unique Characters : {"".join(result)}")
````

### Problem 7: Anagram Check
**Problem statement:** Check whether two strings are anagrams of each other.

**Definition:** Two words are **anagrams** if they contain **exactly the same letters, the same number of times**, in a different order.

**Examples:**
- `listen` and `silent` are anagrams
- `evil` and `vile` are anagrams
- `hello` and `world` are not anagrams
- `rat` and `car` are not anagrams (different letters)

**Approach:**
1. Convert both strings to lowercase.
2. Sort the letters of each string with `sorted()`.
3. If the two sorted lists are equal, the words are anagrams.

**Sample input → output:**
- `Listen, Silent` → `listen & silent are Anagrams`
- `Hello, World` → `hello & world are not Anagrams`

````python
# 7. Check whether two strings are anagrams.
# lower() makes the comparison case-insensitive
word1 = input("Enter the word : ").lower()
word2 = input("Enter the word : ").lower()
# Anagrams have exactly the same letters in a different order, so their sorted letters are identical
# Spaces count as characters; use .replace(' ', '') to compare whole phrases
if sorted(word1) == sorted(word2):
  print(f"{word1} & {word2} are Anagrams")
else:
  print(f"{word1} & {word2} are not Anagrams")
````

### Problem 8: Capitalise Each Word Without `title()`
**Problem statement:** Capitalise the first letter of every word in a sentence without using the `title()` method.

**Definition:** **Capitalising** a word means making its first letter uppercase and the remaining letters lowercase. Here it must be done manually using slicing: `word[0].upper() + word[1:].lower()`.

**Examples:**
- `"hello world"` → `"Hello World"`
- `"pYTHON iS fun"` → `"Python Is Fun"`

**Approach:**
1. Split the sentence into words with `split()`.
2. For each word: first letter `upper()` + the rest `lower()`.
3. Collect the new words in a list and `join()` them with spaces.

**Sample input → output:**
- `hello wORLD python` → `Capitalized String : Hello World Python`

````python
# 8. Capitalise the first letter of each word without using title().
text = input("Enter the word : ")
# split() breaks the sentence into a list of words (extra spaces are ignored)
words = text.split()
# List that will collect the capitalised words
capitalize = []
for word in words:
  # First letter in upper case + the remaining letters in lower case (word[1:] = everything after the first letter)
  new_word = word[0].upper() + word[1:].lower()
  capitalize.append(new_word)
# join() puts the words back together with one space between them
print(f"Capitalized String : {" ".join(capitalize)}")
````

### Problem 9: Split, Join, Replace and Strip
**Problem statement:** Split a sentence into words, join them with `-`, replace the hyphens with underscores, and demonstrate `strip()`, `lstrip()` and `rstrip()`.

**Definition:** `split()` turns a sentence into a list of words; `join()` does the reverse using a chosen separator; `replace(old, new)` substitutes text; `strip()` removes spaces from both ends, `lstrip()` from the left only and `rstrip()` from the right only.

**Examples:**
- `"I love python".split()` → `['I', 'love', 'python']`
- `"-".join(["I", "love"])` → `"I-love"`
- `"a-b".replace("-", "_")` → `"a_b"`
- `"  hi  "` has length 6; after `strip()` the length is 2

**Approach:**
1. Split the sentence and join the words with `-`.
2. Use `replace("-", "_")` on the result.
3. Create a string with spaces on both sides and print its length after each strip method.

**Sample input → output:**
- `I love python` → `I-love-python, then I_love_python, then lengths 20, 5, 15, 10`

````python
# 9. Split a sentence into words, join them with -, and replace spaces with underscores.
# Also demonstrate strip(), lstrip(), and rstrip().
sentence = input("Enter the sentence : ")
# split() with no argument splits on any whitespace and ignores extra spaces
words = sentence.split()
# '-'.join(list) glues the words together using '-' as the separator
result1 = "-".join(words)
print(f"Reshaped Sentence : {result1}")
# replace(old, new) changes every '-' into '_'  (sentence.replace(' ', '_') would do it directly)
print(f"Reshaped Sentence : {result1.replace("-","_")}")
# Demo string: 5 spaces + Hello + 10 spaces
text = "     Hello          "
# Original length: 20 (5 + 5 + 10)
print(len(text))
# strip() removes spaces from BOTH ends: 5
print(len(text.strip()))
# lstrip() removes spaces from the LEFT end only: 15
print(len(text.lstrip()))
# rstrip() removes spaces from the RIGHT end only: 10
print(len(text.rstrip()))
````

### Problem 10: Longest Word in a Sentence
**Problem statement:** Find the longest word in a sentence and print it with its length.

**Definition:** The **longest word** is the word with the greatest number of characters. If two words have the same (greatest) length, the **first** one is reported by this solution.

**Examples:**
- `"I love programming"` → `programming` (11 letters)
- `"cat bat rat"` → `cat` (all have 3 letters, the first wins)

**Approach:**
1. Split the sentence into words.
2. Keep two variables: the longest length so far and the word itself.
3. If a word is longer than the stored length, update both variables.

**Sample input → output:**
- `I love programming` → `The longest word in 'programming' : 11`

````python
# 10. Find the longest word in a sentence.
sen = input("Enter the sentence : ")
words = sen.split()
# Length of the longest word found so far (the name 'max' hides the built-in max())
max = 0
# The longest word found so far
max_word = ""
for word in words:
  # Strictly greater, so when two words have the same length the FIRST one is kept
  if len(word)>max:
    # New longest word: remember both its length and the word
    max = len(word)
    max_word = word
print(f"The longest word in '{max_word}' : {max}")
````

### Problem 11: Toggle the Case Manually
**Problem statement:** Swap the case of every character without using `swapcase()`.

**Definition:** **Toggling case** means every uppercase letter becomes lowercase and every lowercase letter becomes uppercase. Digits, spaces and symbols have no case and stay unchanged.

**Examples:**
- `"Hello World 123"` → `"hELLO wORLD 123"`
- `"PyThOn"` → `"pYtHoN"`

**Approach:**
1. Start with an empty result string.
2. For each character: if it is upper, add `lower()`; if it is lower, add `upper()`; otherwise add it unchanged.
3. Print the result.

**Sample input → output:**
- `Hello World 123` → `Case Swapped String : hELLO wORLD 123`

````python
# 11. This Toggle the case of every character manually, without swapcase().
sen = input("Enter the sentence : ")
# Build the new string one character at a time
res = ""
for letter in sen:
  # Upper -> lower and lower -> upper; digits, spaces and symbols are kept unchanged
  if letter.isupper():
    res += letter.lower()
  elif letter.islower():
    res += letter.upper()
  else:
    # Not a letter (digit, space or symbol): keep it as it is
    res += letter
print(f"Case Swapped String : {res}")
````

### Problem 12: Pangram
**Problem statement:** Check whether a sentence is a pangram.

**Definition:** A **pangram** is a sentence that contains **every letter of the English alphabet at least once**. The classic example is "The quick brown fox jumps over the lazy dog".

**Examples:**
- `"The quick brown fox jumps over the lazy dog"` is a pangram
- `"Pack my box with five dozen liquor jugs"` is a pangram
- `"Hello world"` is not a pangram (many letters are missing)

**Approach:**
1. Convert the sentence to lowercase.
2. Assume `pangram = True`.
3. Check each letter from `a` to `z`; if one is missing, set `pangram = False` and stop.

**Sample input → output:**
- `The quick brown fox jumps over the lazy dog` → `The sentence is a pangram.`
- `Hello world` → `The sentence is not a pangram.`

````python
# 12. Check whether a sentence is a pangram.
sentence = input("Enter a sentence : ").lower()
# A pangram uses every letter from a to z at least once
alphabet = "abcdefghijklmnopqrstuvwxyz"
# Flag: assume it is a pangram until a missing letter is found
pangram = True

# Check all 26 letters one by one
for letter in alphabet:
  if letter not in sentence:
    # One missing letter is enough to fail, so stop checking
    pangram = False
    break

if pangram:
    print("The sentence is a pangram.")
else:
    print("The sentence is not a pangram.")
````

### Problem 13: Caesar Cipher
**Problem statement:** Encrypt a message with a given shift, then decrypt it back.

**Definition:** A **Caesar cipher** is a simple encryption technique in which every letter is moved a fixed number of places (the **shift**) forward in the alphabet, wrapping around from `z` back to `a`. Decrypting moves each letter the same number of places **backward**. Non-letters are left unchanged.

**Examples:**
- Shift 3: `A → D`, `B → E`, `X → A` (wraps around)
- `"Hello, World!"` with shift 3 → `"Khoor, Zruog!"`
- The case of each letter is preserved; the comma, space and `!` are unchanged

**Approach:**
1. Convert a letter to a number 0-25 with `ord(ch) - 65` (uppercase) or `ord(ch) - 97` (lowercase).
2. Add the shift and use `% 26` to wrap around.
3. Convert back with `chr(... + 65)` or `chr(... + 97)`.
4. Decrypt with the same formula using `- shift`.

**Sample input → output:**
- `Hello, World! with shift 3` → `Encrypted: Khoor, Zruog! then Decrypted: Hello, World!`

````python
# 13. Implement a Caesar cipher (encrypt and decrypt) with a given shift.
text = input("Enter a message: ")
# Caesar cipher: every letter moves 'shift' places forward in the alphabet and wraps around after z
shift = int(input("Enter shift value: "))

# Encrypt
encrypted = ""
for char in text:
    # ord() gives the letter's code; subtracting 65 ('A') turns A-Z into 0-25
    # add the shift, % 26 wraps around the alphabet, then + 65 and chr() turn it back into a letter
    if char.isupper():
        encrypted += chr((ord(char) - 65 + shift) % 26 + 65)
    # Lowercase letters work the same way, using 97 ('a') as the starting point
    elif char.islower():
        encrypted += chr((ord(char) - 97 + shift) % 26 + 97)
    else:
        # Spaces, digits and punctuation are not changed
        encrypted += char

print("Encrypted:", encrypted)

# Decrypt (shift backward)
decrypted = ""
for char in encrypted:
    # Same formula, but the shift is SUBTRACTED to undo the encryption
    if char.isupper():
        decrypted += chr((ord(char) - 65 - shift) % 26 + 65)
    elif char.islower():
        decrypted += chr((ord(char) - 97 - shift) % 26 + 97)
    else:
        decrypted += char

print("Decrypted:", decrypted)
````

### Problem 14: String Compression (Run-Length Encoding)
**Problem statement:** Compress a string so that `"aaabbc"` becomes `"a3b2c1"`.

**Definition:** **Run-length encoding** replaces every group of identical consecutive characters with the character followed by the length of the group.

**Examples:**
- `"aaabbc"` → `"a3b2c1"`
- `"aaaa"` → `"a4"`
- `"abc"` → `"a1b1c1"` (no repeats, so the result is actually longer)

**Approach:**
1. Compare each character with the one before it.
2. If it is the same, increase `count`.
3. If it is different, store the previous character and its count, then reset `count = 1`.
4. After the loop, store the last group.

**Sample input → output:**
- `aaabbc` → `Compressed: a3b2c1`

````python
# 14. Compress a string: "aaabbc" becomes "a3b2c1".
# Run-length encoding: replace each run of repeated characters with the character and its count
text = input("Enter a string to compress: ")

# An empty string has nothing to compress
if not text:
    print("Compressed: ")
else:
    compressed = ""
    # count = length of the current run of identical characters
    count = 1

    # Compare each character with the one before it
    for i in range(1, len(text)):
        # Same as the previous character -> the run continues
        if text[i] == text[i - 1]:
            count += 1
        else:
            # The run ended: store the character and its count, then restart the count
            compressed += text[i - 1] + str(count)
            count = 1

    # Add the final character group
    compressed += text[-1] + str(count)
    # Note: a string with no repeats gets longer ('abc' becomes 'a1b1c1')
    print("Compressed:", compressed)
````

### Problem 15: Extract and Validate an Email Address
**Problem statement:** Extract the username and domain from an email address and check its basic format using only string methods.

**Definition:** An email address has the form `username@domain`. A **basic** validity check requires: no spaces, exactly one `@`, a non-empty username, and a domain that contains a `.` which is not at the start or the end (for example `example.com`). Real-world email validation has many more rules, so this is only a simple check.

**Examples:**
- `user@example.com` → valid (username `user`, domain `example.com`)
- `user@@example.com` → invalid (two `@`)
- `user@example` → invalid (no `.` in the domain)
- `@example.com` → invalid (empty username)

**Approach:**
1. Remove surrounding spaces with `strip()`.
2. Check that there is no space and `count("@") == 1`.
3. Split at `@` into username and domain.
4. Check that the username is not empty and the domain has a valid `.`.

**Sample input → output:**
- `user@example.com` → `Valid email! / Username: user / Domain: example.com`
- `user@@example.com` → `Invalid email format.`

````python
# 15. Extract the username and domain from an email address, and validate its basic format using only string methods.
# strip() removes accidental spaces at the start and the end
email = input("Enter an email address: ").strip()

# Basic validation using only string methods
# Rule 1: the address must not contain a space
has_no_spaces = " " not in email
# Rule 2: there must be exactly one '@' symbol
has_one_at = email.count("@") == 1

# Split only when both basic rules pass; otherwise the address is invalid
if has_no_spaces and has_one_at:
    # split('@') gives two parts: the text before '@' is the username and the text after it is the domain
    username, domain = email.split("@")

    # Domain must contain at least one '.', not start/end with '.', and username must not be empty
    valid_username = len(username) > 0
    valid_domain = (
        "." in domain
        and not domain.startswith(".")
        and not domain.endswith(".")
        # The shortest possible domain has 3 characters, such as 'a.b'
        and len(domain) >= 3
    )

    # Both parts must be valid for the address to be accepted
    if valid_username and valid_domain:
        print("Valid email!")
        print("Username:", username)
        print("Domain:", domain)
    else:
        print("Invalid email format.")
else:
    print("Invalid email format.")
````
