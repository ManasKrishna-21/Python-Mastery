# Random Library

## Description
The **`random` module** is part of Python's standard library, so it needs no installation. You only have to import it once at the top of your program:

```python
import random
```

Computers cannot produce truly random numbers by themselves. Python uses a mathematical formula that creates numbers that *look* random. This is called **pseudo-random** generation. The formula starts from a **seed** value, so using the same seed always gives the same sequence. This is very useful for testing and debugging.

### Commonly used functions
| Function | What it does | Example |
|----------|--------------|---------|
| `random.random()` | Random float from `0.0` up to (but not including) `1.0` | `0.7364` |
| `random.randint(a, b)` | Random integer from `a` to `b`, **both included** | `randint(1, 6)` → `1` to `6` |
| `random.uniform(a, b)` | Random float between `a` and `b` | `uniform(5, 10)` → `7.31` |
| `random.randrange(start, stop, step)` | Like `range()`, but picks one value (stop **excluded**) | `randrange(0, 10, 2)` → `0, 2, 4, 6, 8` |
| `random.choice(seq)` | Picks **one** item from a sequence | `choice(["a", "b", "c"])` |
| `random.choices(seq, k=n)` | Picks `n` items **with replacement** (repeats possible) | `choices("HALO", k=3)` |
| `random.sample(seq, k)` | Picks `k` **unique** items (no repeats) | `sample(range(1, 50), 6)` |
| `random.shuffle(list)` | Rearranges a list **in place** | `shuffle(cards)` |
| `random.seed(n)` | Fixes the starting point so results can be repeated | `seed(42)` |

### Key points
- `randint(a, b)` includes **both** ends, but `range()` and `randrange()` stop **before** the end value.
- `choices()` can return the same item twice; `sample()` never does. `sample()` raises an error if `k` is larger than the sequence.
- `shuffle()` changes the original list and returns `None`, so do not write `cards = random.shuffle(cards)`.
- `choice()`, `choices()` and `sample()` work on any sequence: lists, strings, tuples and `range` objects.
- The `string` module gives ready-made character sets: `string.ascii_letters`, `string.digits`, `string.punctuation`.
- `random` is **not secure**. For real passwords or tokens use the `secrets` module instead.

````python
# 8. Use random.seed() to show that the same seed gives the same “random” numbers.
import random
# seed() sets the starting point of the random generator: the same seed always gives the same sequence
# This is why the numbers are called pseudo-random (they come from a fixed formula)
random.seed(42)
draw_1 = random.sample(range(1, 50), 6)
# Resetting to seed 42 repeats exactly the same draw
random.seed(42)
draw_2 = random.sample(range(1, 50), 6)
# A different seed gives a different draw
random.seed(99)
draw_3 = random.sample(range(1, 50), 6)

# draw_1 and draw_2 are identical, while draw_3 is different
print(f"Seed 42 (Run 1): {draw_1}")
print(f"Seed 42 (Run 2): {draw_2}")
print(f"Seed 99 (Run 3): {draw_3}")
````

````python
# 9. Generate a random 8-character password with letters, digits, and symbols (use the string module).
import random
import string

# ascii_letters = a-z and A-Z, digits = 0-9, punctuation = symbols such as ! @ # $ %
characters = string.ascii_letters + string.digits + string.punctuation
# choices(..., k=8) picks 8 characters (repeats allowed); join() glues them into one string
# For real passwords prefer the 'secrets' module, which is designed for security
password = "".join(random.choices(characters, k=8))

print(f"Generated password: {password}")
````

````python
# 10. Rock-Paper-Scissors against the computer, best of 5, with score tracking.
import random

# The three valid moves; the computer picks one of them at random
choices = ["rock", "paper", "scissors"]
user_score = 0
computer_score = 0
target_wins = 3  # First to 3 wins in a best-of-5 match

print("Rock-Paper-Scissors: Best of 5 (First to 3 wins!)")

# Keep playing until either player reaches 3 wins
while user_score < target_wins and computer_score < target_wins:
  # strip() removes extra spaces and lower() makes the input case-insensitive
  user_choice = input("\nEnter rock, paper, or scissors: ").strip().lower()

  # Validate the input; 'continue' skips the rest of this round and asks again
  if user_choice not in choices:
    print("Invalid choice. Please type rock, paper, or scissors.")
    continue

  # The computer picks a random move
  computer_choice = random.choice(choices)
  print(f"Computer chose: {computer_choice}")

  # Same move on both sides -> tie (no score change)
  if user_choice == computer_choice:
    print("It's a tie!")
  # The user wins if: rock beats scissors, paper beats rock, or scissors beats paper
  elif (
      (user_choice == "rock" and computer_choice == "scissors")
      or (user_choice == "paper" and computer_choice == "rock")
      or (user_choice == "scissors" and computer_choice == "paper")
  ):
    user_score += 1
    print("You win this round!")
  else:
    # Any other combination means the computer wins this round
    computer_score += 1
    print("Computer wins this round!")

  print(f"Score -> You: {user_score} | Computer: {computer_score}")

# Announce the overall winner
print("\n--- Game Over ---")
if user_score == target_wins:
  print("Congratulations! You won the best-of-5 series!")
else:
  print("Computer won the best-of-5 series. Better luck next time!")
````

# While Loops

## Description
A **`while` loop** repeats a block of code **as long as its condition stays `True`**. It is the best choice when you do *not* know in advance how many times the loop must run, for example "keep asking until the password is correct".

### Syntax
```python
while condition:
    # code to repeat (indented)
    # update something so the condition can eventually become False
```

### How it works
1. Python checks the condition.
2. If it is `True`, the block runs, then Python goes back to step 1.
3. If it is `False`, the loop ends and the program continues after it.

### The three parts of a good while loop
| Part | Purpose | Example |
|------|---------|---------|
| **Initialise** | Set the starting value before the loop | `num = 1` |
| **Condition** | Decides whether to continue | `while num <= 10:` |
| **Update** | Changes the value so the loop can end | `num += 1` |

If you forget the update, the condition never becomes `False` and you get an **infinite loop**. (In a notebook, press the stop button to interrupt it.)

### `for` loop vs `while` loop
| | `for` loop | `while` loop |
|--|-----------|--------------|
| Use when | You know the sequence or the number of rounds | You only know the stopping condition |
| Example | Print 1 to 10 | Repeat until the user types "quit" |

### Controlling a while loop
| Statement | Effect |
|-----------|--------|
| `break` | Exit the loop immediately |
| `continue` | Skip to the next round (the condition is checked again) |
| `else` | Runs when the loop ends normally (not by `break`) |
| `while True:` | Intentional infinite loop, stopped with `break` |

### Digit-handling toolkit
Many number problems repeat the same two steps until the number becomes `0`:

| Operation | Meaning | Example (`num = 1234`) |
|-----------|---------|------------------------|
| `num % 10` | Get the **last digit** | `4` |
| `num // 10` | **Remove** the last digit | `123` |
| `rev * 10 + digit` | Build a reversed number | `0 → 4 → 43 → 432 → 4321` |
| `num % 2` and `num // 2` | Convert to binary | remainders read from bottom to top |

### Common patterns
- **Counter loop:** counts from a start value to a limit.
- **Accumulator:** collects a running total, as in the digit sum.
- **Sentinel loop:** runs until a special value is entered, for example a negative number.
- **Retry loop:** limits the number of attempts, as in the password example.
- **Sequence generator:** updates values each round, as in Collatz and Fibonacci.

### Key points
- Always make sure something inside the loop moves the condition towards `False`.
- Keep a copy (`temp = num`) if you need the original value after the loop changes it.
- Avoid naming a variable `sum`, because it hides the built-in `sum()` function.

````python
# 1. Print numbers 1 to 10 using a while loop.
# Initialise the counter before the loop starts
num = 1
# The loop continues as long as this condition is True
while num<=10:
  print(num, end = " ")
  # Update the counter; without this the condition never becomes False (infinite loop)
  num += 1
````

````python
# 2. Find the sum of the digits of a number.
num = int(input("Enter the number : "))
# Keep a copy of the original number because num shrinks inside the loop
temp = num
# Accumulator for the total of the digits (the name 'sum' hides Python's built-in sum())
sum = 0
while num > 0:
  # num % 10 gives the last digit (for example 1234 % 10 = 4)
  rem = num%10
  # Add that digit to the total
  sum += rem
  # num // 10 removes the last digit (1234 -> 123); the loop ends when num becomes 0
  num = num//10
print(f"Sum of Digits for {temp} : {sum}")
````

````python
# 3. Reverse a number.
num = int(input("Enter the number : "))
temp = num
# Will hold the reversed number (the name 'sum' hides Python's built-in sum())
sum = 0
while num > 0:
  # Take the last digit
  rem = num%10
  # Shift the reversed number one place left (x10) and add the new digit: 0 -> 4 -> 43 -> 432 -> 4321
  sum = (sum*10) + rem
  # Remove the last digit
  num = num//10
print(f"Reverse of Number {temp} : {sum}")
````

````python
# 4. Count the digits in a number.
num = int(input("Enter the number : "))
temp = num
# Counter for the number of digits
count = 0
while num > 0:
  # Remove one digit each round ...
  num = num//10
  # ... and count it (for the number 0 the loop never runs, so it would print 0 digits)
  count += 1
print(f"Count of Digits {temp} : {count}")
````

````python
# 5. Keep asking for a password until the correct one is entered, with a 3-attempt limit.
# The correct password (comparison is case-sensitive)
password = "Manas"
# count tracks the number of wrong attempts
count = 0
# Allow at most 3 attempts
while count<3:
  trial = input("Enter the password : ")
  if password == trial:
    print("Correct Password, Welcome back, Manas")
    # break leaves the loop immediately after a correct password
    break
  else:
    # Wrong password: record one more failed attempt
    count+=1
    print("Invalid Password")
    # After the third wrong attempt the loop condition becomes False, so show the final message
    if count == 3:
      print("All tries exhausted")

````

````python
# 6. Print the Collatz sequence for a starting number until it reaches 1.
# Collatz rule: if the number is even divide it by 2, if odd multiply by 3 and add 1
# The sequence is believed to always reach 1
num = int(input("Enter a positive starting number : "))
print("Collatz sequence :", end = " ")
# Repeat until the sequence reaches 1
while num != 1:
  print(f"{num} -->",end = " ")
  # Even number -> halve it
  if num%2 == 0:
    # '/' gives a float, so int() converts it back to a whole number (num // 2 does this directly)
    num = int(num/2)
  else:
    # Odd number -> multiply by 3 and add 1
    num = (3*num)+1
# The loop stops before printing 1, so print it here to complete the sequence
print(1)
````

````python
# 7. Keep adding numbers until the user enters a negative number, then print the total and the average.
# count = how many numbers were entered, sum = their total
count = 0
sum = 0
# Intentional infinite loop; it is stopped with break when the user enters a non-positive number
while True:
  num = int(input("Enter a positive numbers : "))
  # Positive number -> include it in the count and the total
  if num >0:
    count += 1
    sum += num
  else:
    # Zero or negative number -> stop the loop (this value is the 'sentinel')
    break
# Average = total / count (this fails with ZeroDivisionError if no positive number was entered)
print(f"Sum of the entered {count} digits : {sum}\nAverage of entered {count} digits : {round(sum/count,2)}")
````

````python
# 8. Convert a decimal number to binary using a while loop (no bin()).
num = int(input("Enter a non-negative decimal number: "))
# Keep the original number for the final message
temp = num
binary = ""

# Special case: the loop below never runs for 0, so handle it separately
if num == 0:
  binary = "0"
else:
  while num > 0:
    # The remainder after dividing by 2 is the next binary digit (starting from the rightmost)
    remainder = num % 2
    binary = str(remainder) + binary  # Prepend remainder to build right-to-left
    # Divide by 2 and repeat until the number becomes 0. Example: 10 -> 5 -> 2 -> 1 -> 0 gives 1010
    num = num // 2

print(f"Binary of {temp} is: {binary}")
````

````python
# 9. Print all Fibonacci numbers below 1000.
# a and b are two consecutive terms of the series (0, 1, 1, 2, 3, 5, 8 ...)
a = 0
b = 1
print("Fibonacci Series :",end = " ")
# Keep going while the current term is below 1000
while a < 1000:
  print(a,end = " ")
  # Move forward: a takes the next term and b takes the sum of the two (both updates happen together)
  a,b = b,a+b
````

````python
# 10. Find the digital root of a number by repeatedly summing digits until one digit remains
#(for example, 9875 → 29 → 11 → 2)
num = int(input("Enter a positive integer : "))
# Keep the original number for the final message
temp = num

# Outer loop: repeat while the number still has more than one digit
while num >= 10:
  # Reset the sum for each pass
  digit_sum = 0
  # Inner loop: add up all the digits of the current number
  while num > 0:
    digit_sum += num % 10
    num = num // 10
  # The digit sum becomes the new number (9875 -> 29 -> 11 -> 2)
  num = digit_sum

# Shortcut for a positive number: digital root = 1 + (n - 1) % 9
print(f"Digital root of {temp} : {num}")
````
