# Conditional Statements

## Description
A **conditional statement** lets a program choose between different paths based on whether a condition is `True` or `False`. A *condition* is any expression that produces a boolean, such as `age >= 18` or `x == 5`.

### Syntax
```python
if condition1:
    # runs when condition1 is True
elif condition2:
    # runs when condition1 is False and condition2 is True
else:
    # runs when none of the conditions above are True
```

| Keyword | Purpose | Required? |
|---------|---------|-----------|
| `if` | First condition to test | Yes |
| `elif` | Extra conditions (can be used many times) | No |
| `else` | Fallback when nothing matched (has no condition) | No |

### How it works
1. Python checks the conditions **from top to bottom**.
2. The **first** condition that is `True` runs its block.
3. All remaining `elif` / `else` blocks are **skipped**.
4. If nothing is `True`, the `else` block runs (if present).

### Key points
- A colon `:` must follow every `if`, `elif` and `else`.
- The lines inside a block must be **indented**.
- Use `==` to **compare** and `=` to **assign**. Writing `if x = 5:` is an error.
- Combine conditions with `and`, `or`, `not`. Chained comparisons such as `1 <= x <= 10` are allowed.
- **Order matters** in an `if-elif` chain: put the most specific / highest condition first.
- An `if` placed inside another `if` is called a **nested if**.
- Empty values (`0`, `""`, `[]`, `None`) are treated as `False`; everything else is `True`.

### Useful tools inside conditions
| Tool | Example | Meaning |
|------|---------|---------|
| `%` remainder | `n % 2 == 0` | n is even |
| `in` | `ch in "aeiou"` | ch is one of the vowels |
| `isalpha()` | `"a".isalpha()` | is a letter |
| `isdigit()` | `"7".isdigit()` | is a digit |
| `isupper()` / `islower()` | `"A".isupper()` | is uppercase / lowercase |
| `lower()` | `"A".lower()` | makes comparison case-insensitive |

````python
# 1. Check whether a number is positive, negative, or zero.
# input() returns text, so int() converts it into a whole number
num = int(input("Enter the number : "))
# Python checks the conditions from top to bottom and runs only the FIRST one that is True
if(num > 0):
  print(f"{num} is a Positive Number")
elif(num == 0):
  # num == 0 (a double '==' compares, a single '=' would assign)
  print(f"{num} is a Zero")
else:
  # else has no condition: it runs when all the conditions above are False (so num < 0)
  print(f"{num} is a Negative Number")
````

````python
# 2. Check whether a number is even or odd.
# An even number leaves remainder 0 when divided by 2
num = int(input("Enter the number : "))
if(num%2==0):
  print(f"{num} is an Even Number")
else:
  # Otherwise the remainder is 1, so the number is odd
  print(f"{num} is an Odd Number")
````

````python
# 3. Find the largest of three numbers.
num1 = int(input("Enter the number 1 : "))
num2 = int(input("Enter the number 2 : "))
num3 = int(input("Enter the number 3 : "))
# 'and' needs BOTH comparisons to be True: num1 must be greater than num2 and num3
if(num1 > num2 and num1 > num3):
  print(f"{num1} is the largest number")
elif(num2 > num3):
  # num1 is already ruled out, so the largest is num2 if it beats num3
  print(f"{num2} is the largest number")
else:
  # Otherwise num3 is the largest (or tied for largest)
  print(f"{num3} is the largest number")
````

````python
# 4. Check whether a year is a leap year.
year = int(input("Enter the year : "))
# Leap year rule: divisible by 4, EXCEPT century years, which must also be divisible by 400
if(year % 4 == 0):
    # Century year (1900, 2000, ...): needs the extra check by 400
    if(year % 100 == 0):
        # 2000 is a leap year (divisible by 400), but 1900 is not
        if(year % 400 == 0):
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        # Divisible by 4 but not by 100 (for example 2024) -> leap year
        print(f"{year} is a leap year")
else:
    # Not divisible by 4 -> not a leap year
    print(f"{year} is not a leap year")
````

````python
# 5. Convert marks to a grade (A/B/C/D/F) using if-elif-else.
# Scale used here: 90+ S, 80+ A, 70+ B, 60+ C, 50+ D, 40+ E, below 40 F
grade = int(input("Enter your grade in the subject : "))
# Check from the highest mark downwards, so each elif only needs a lower limit
if(grade>=90):
  print("Grade : S")
elif(grade>=80):
  print("Grade : A")
elif(grade>=70):
  print("Grade : B")
elif(grade>=60):
  print("Grade : C")
elif(grade>=50):
  print("Grade : D")
elif(grade>=40):
  print("Grade : E")
else:
  # Below 40 marks -> fail
  print("Grade : F")
````

````python
# 6. Check whether three sides form a valid triangle, and if so, whether it is equilateral, isosceles, or scalene.
side1 = int(input("Enter the side 1 : "))
side2 = int(input("Enter the side 2 : "))
side3 = int(input("Enter the side 3 : "))
# Valid triangle rule: the sum of any two sides must be greater than the third side
# Chained comparison below is True only when all three sides are equal
if(side1 == side2 == side3):
  print("Equilateral Triangle")
# Exactly two equal sides (all three equal was already handled above)
elif(side1 == side2 or side2 == side3 or side1 == side3):
  print("Isosceles Triangle")
else:
  # All three sides are different
  print("Scalene Triangle")
````

````python
# 7. Check whether a letter is a vowel or a consonant.
char = input("Enter the character : ")
# lower() makes the check case-insensitive, so 'A' and 'a' are treated the same
char = char.lower()
# List of vowels to compare against
vowels = ['a','e','i','o','u']
# 'in' checks whether the character is present in the list
# Note: any input that is not a vowel (even a digit or symbol) is reported as a consonant
if char in vowels:
  print(f"{char} is a vowel")
else:
  print(f"{char} is a consonant")
````

````python
# 8. Electricity bill: calculate the amount from slab rates
# (for example, the first 100 units at one rate, the next 100 at another).
# Fixed bill per slab: up to 50 units -> 55$, 51-100 -> 110$, 101-150 -> 165$, above 150 -> 220$
slab = int(input("Enter the slab of the month : "))
# '<=' means less than or equal to; each elif continues from where the previous range ended
if(slab<=50):
  print("Bill : 55$")
elif(slab<=100):
  print("Bill : 110$")
elif(slab<=150):
  print("Bill : 165$")
else:
  print("Bill : 220$")
````

````python
# 9. Build a simple calculator (+ - * / %) using if-elif, handling division by zero.
num1 = int(input("Enter the number 1 : "))
num2 = int(input("Enter the number 2 : "))
# Supported operators: + - x /   (multiplication is typed as the letter 'x')
op = input("Enter the operation : ")
# Compare the operator typed by the user with each supported symbol
if(op == "+"):
  print(f"{num1} {op} {num2} = {num1 + num2}")
elif(op == "-"):
  print(f"{num1} {op} {num2} = {num1 - num2}")
elif(op == "x"):
  print(f"{num1} {op} {num2} = {num1 * num2}")
elif(op == "/"):
  # Dividing by zero raises an error, so check that the divisor is not 0 first
  if(num2 != 0):
    print(f"{num1} {op} {num2} = {num1 / num2}")
  else:
    # Divisor is 0 -> show a message instead of crashing
    print("Opertion not possible")
else:
  # Any other symbol falls through to here (for example %, which is not handled)
  print("Invalid Operation")
````

````python
# 10. Classify a single character as uppercase, lowercase, digit, or special character.
char = input("Enter the character : ")
# isalpha() is True for letters (A-Z, a-z)
if char.isalpha():
  # A letter is either uppercase or lowercase
  if char.isupper():
    print(f"{char} is uppercase")
  else:
    print(f"{char} is lowercase")
# isdigit() is True for the characters 0-9
elif char.isdigit():
  print(f"{char} is digit")
else:
  # Not a letter and not a digit -> special character such as @ # $ or a space
  print(f"{char} is special character")
````

# For Loops


## Description
A **loop** repeats a block of code. A **`for` loop** is used when you know *what* to loop over or *how many times* to repeat. It takes each item of a sequence (a range of numbers, a string, a list...) one by one and runs the indented block for it.

### Syntax
```python
for variable in sequence:
    # code to repeat (indented)
```

### The `range()` function
`range()` produces a sequence of integers. The **stop value is never included**.

| Form | Meaning | Example | Values produced |
|------|---------|---------|-----------------|
| `range(stop)` | 0 up to stop-1 | `range(5)` | 0, 1, 2, 3, 4 |
| `range(start, stop)` | start up to stop-1 | `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(start, stop, step)` | jump by `step` | `range(2, 11, 3)` | 2, 5, 8 |
| negative step | count down | `range(5, 0, -1)` | 5, 4, 3, 2, 1 |

### Looping over other sequences
- A **string** gives one character at a time.
- A **list** gives one element at a time.
- `enumerate(seq)` gives `(index, item)` pairs, so you get the position and the value together.

### Common loop patterns
| Pattern | Idea | Typical code |
|---------|------|--------------|
| **Accumulator (sum)** | start at `0`, keep adding | `total = 0` then `total += i` |
| **Accumulator (product)** | start at `1`, keep multiplying | `fac = 1` then `fac *= i` |
| **Counting** | add 1 when something matches | `count += 1` |
| **Series** | update values each round | Fibonacci: `a, b = b, a + b` |
| **Nested loops** | loop inside a loop | rows and columns for patterns |

### Controlling a loop
| Statement | Effect |
|-----------|--------|
| `break` | Stop the loop immediately |
| `continue` | Skip the rest of this round and go to the next item |
| `else` (on a loop) | Runs only if the loop finished **without** a `break` |

### Key points
- Use `print(..., end=" ")` to print on the same line, and an empty `print()` to move to the next line.
- In **nested loops**, the inner loop runs fully for every single round of the outer loop. This is how patterns are drawn: the outer loop handles rows, the inner loop handles columns.
- Avoid naming a variable `sum`, `max` or `min`, because it hides the built-in function of the same name.

````python
# 1. Print numbers 1 to 10.
# range(1, 11) gives 1, 2, ... 10 (the stop value 11 is NOT included)
for i in range(1,11):
  # end=" " keeps all numbers on one line, separated by spaces
  print(i,end = " ")
````

````python
# 2. Print the multiplication table of a number entered by the user.
num = int(input("Enter the number for multiplication table : "))
# i runs from 1 to 10; each round prints num multiplied by i
for i in range(1,11):
  print(f"{num} x {i} = {num*i}")
````

````python
# 3. Find the sum of the first n natural numbers.
num = int(input("Enter the number : "))
# Accumulator: starts at 0 and collects the running total
# Note: the name 'sum' hides Python's built-in sum() function; a name like total is safer
sum = 0
# num+1 is used because range() stops BEFORE the end value
for i in range(1,num+1):
  # Add the current number to the running total (same as sum = sum + i)
  sum += i
# Check with the formula: n * (n + 1) / 2
print(f"Sum of the first {num} : {sum}")
````

````python
# 4. Find the factorial of a number.
num = int(input("Enter the number : "))
# Factorial is a product, so start at 1 (starting at 0 would always give 0)
fac = 1
for i in range(1,num+1):
  # n! = 1 x 2 x 3 x ... x n; each round multiplies the result so far by i
  # For 0 the loop does not run, so the result stays 1 (0! is defined as 1)
  fac *= i
print(f"Factorial of {num} : {fac}")
````

````python
# 5. Print all even numbers between 1 and 50 using the range() step.
# Alternative: range(2, 51, 2) would produce only even numbers using the step value
print("Even Numbers : ",end = " ")
for i in range(1,51):
  # Remainder 0 when divided by 2 means the number is even
  if i%2==0:
    print(i,end = " ")
````

````python
# 6. Print a right-angled triangle pattern of stars for n rows.
rows = int(input("Enter the number of rows : "))
# Outer loop controls the rows. Row number i+1 prints i+1 stars
for i in range(rows):
  # Inner loop prints the stars of the current row (1 star, then 2, then 3 ...)
  for j in range(i+1):
    print("*",end = " ")
  # An empty print() moves the cursor to the next line after each row
  print()
````

````python
# 7. Print a pyramid pattern of stars for n rows.
rows = int(input("Enter the row size for the pattern: "))
# Outer loop: one round per row. Each row = leading spaces + an odd number of stars (1, 3, 5 ...)
for i in range(rows):
  # First inner loop prints the leading spaces (fewer spaces on the lower rows)
  for j in range(i,rows):
    print(" ",end = " ")
  # Second inner loop prints 2*i + 1 stars, because range(1, 2*(i+1)) has that many values
  for j in range(1,2*(i+1)):
    print("*",end = " ")
  # Move to the next row
  print()
````

````python
# 8. Print the first n terms of the Fibonacci series.
# Fibonacci series: every term is the sum of the two terms before it (0, 1, 1, 2, 3, 5 ...)
a = 0
b = 1
num = int(input("Enter the no of terms in Fibonacci Series : "))
# The first two terms are printed directly (this assumes at least 2 terms are requested)
print(f"{a} {b}",end = " ")
# Only num-2 more terms are needed because 2 terms are already printed
for i in range(num-2):
  # Next term = sum of the previous two terms
  c = a + b
  print(c,end = " ")
  # Slide forward: b becomes the older term and c becomes the newer one
  a,b = b,c
````

````python
# 9. Print 10 down to 1, then loop over a string and print each character with its index.
# range(start, stop, step): the step -1 counts down from 10 to 1 (stop value 0 is not included)
for i in range(10,0,-1):
  print(f"{i}",end = " ")
print()
# A string can be looped over one character at a time
text = "Python"
# enumerate() gives (index, character) pairs, so no manual counter is needed
for index, char in enumerate(text):
  # index counts from the left (0 to 5); -(len(text) - index) is the same position from the right (-6 to -1)
  print(f"Index {index} / -{len(text)-index} : {char}")
````

````python
# 10. Print this number pattern for n = 5: 1, 12, 123, 1234, 12345.
# Build the string step by step: each row is the previous row plus the next number
text = ""
# i goes from 1 to 5 (change 6 to n+1 to work for any n)
for i in range(1,6):
  # str(i) turns the number into text so it can be joined to the string
  text += str(i)
  # Prints 1, 12, 123, 1234, 12345 on separate lines
  print(text)
````
