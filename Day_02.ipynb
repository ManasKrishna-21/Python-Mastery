# Basic Data Types

## Description
Every value in Python has a **data type**. The type decides what the value represents, how much memory it uses and which operations are allowed. Python is **dynamically typed**, so you never declare a type; Python infers it from the value.

| Type | Class name | Meaning | Example |
|------|-----------|---------|---------|
| Integer | `int` | Whole numbers of any size | `42`, `-7`, `0` |
| Floating point | `float` | Numbers with a decimal point | `3.14`, `2.0`, `-0.5` |
| Complex | `complex` | Real + imaginary part (uses `j`) | `3 + 4j` |
| String | `str` | Text | `"Python"`, `'A'` |
| Boolean | `bool` | Logical value | `True`, `False` |
| NoneType | `NoneType` | "No value" placeholder | `None` |

### 1. Checking types: `type()` and `isinstance()`
- `type(x)` tells you the exact class of `x`.
- `isinstance(x, T)` answers *"is `x` of type `T`?"* with `True` or `False`. It is the preferred way to check types in real programs.
- `bool` is a **subclass of `int`**, so `True` behaves like `1` and `False` like `0`.

### 2. Type conversion (casting)
Converting a value from one type to another is done with `int()`, `float()`, `str()`, `bool()`, `complex()`.

| Conversion | Result | Note |
|-----------|--------|------|
| `int("123")` | `123` | string must look like a whole number |
| `float(45)` | `45.0` | |
| `int(3.99)` | `3` | **truncates**, it does not round |
| `bool(0)` | `False` | |
| `str(77)` | `'77'` | |
| `int("12.5")` | `ValueError` | cannot convert a decimal string directly |

### 3. Division and exponent behaviour
- `/` always returns a **float** (`5 / 2 = 2.5`).
- `//` is **floor division**: it rounds *down* (`5 // 2 = 2`, `-7 // 2 = -4`).
- A negative exponent gives a float (`2 ** -1 = 0.5`).
- `True + 1` is `2` because `True` is treated as `1`.

### 4. Floating-point precision
Computers store floats in binary, and numbers like `0.1` cannot be stored exactly. So `0.1 + 0.2` is `0.30000000000000004`. Use `round()` or `math.isclose()` to compare floats safely.

### 5. Complex numbers
A complex number has a real and an imaginary part. Use `.real`, `.imag` and `abs(z)` (the magnitude).

### 6. Truthy and falsy values
Every object can be used as a condition. These are **falsy** (treated as `False`): `0`, `0.0`, `""`, `[]`, `{}`, `()`, `None`, `False`. Everything else is **truthy**, including `" "` (a space) and the text `"False"`.

### 7. Number systems
| Function | Converts to | Prefix |
|----------|-------------|--------|
| `bin(n)` | Binary (base 2) | `0b` |
| `oct(n)` | Octal (base 8) | `0o` |
| `hex(n)` | Hexadecimal (base 16) | `0x` |
| `int(text, base)` | Back to decimal | |

### 8. Big integers
Python's `int` has **unlimited precision**. It grows in memory as needed, so there is no integer overflow like in C or Java.

### 9. Characters and codes
`ord(ch)` gives the numeric code of a character; `chr(code)` gives the character for a code. For example `ord("A") = 65`, `chr(97) = 'a'`.

# Operators & Their Types

## Description
An **operator** is a symbol that performs an operation on one or more values, called **operands**. In `5 + 3`, the operator is `+` and the operands are `5` and `3`. Python groups its operators into several families.

### 1. Arithmetic operators
| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `17 + 5` | `22` |
| `-` | Subtraction | `17 - 5` | `12` |
| `*` | Multiplication | `17 * 5` | `85` |
| `/` | True division (float) | `17 / 5` | `3.4` |
| `//` | Floor division | `17 // 5` | `3` |
| `%` | Modulus (remainder) | `17 % 5` | `2` |
| `**` | Exponent | `2 ** 10` | `1024` |

### 2. Comparison (relational) operators
They compare two values and return `True` or `False`.

| Operator | Meaning | Example (`a=10, b=20`) |
|----------|---------|------------------------|
| `==` | Equal to | `a == b` → `False` |
| `!=` | Not equal to | `a != b` → `True` |
| `>` | Greater than | `a > b` → `False` |
| `<` | Less than | `a < b` → `True` |
| `>=` | Greater than or equal | `a >= b` → `False` |
| `<=` | Less than or equal | `a <= b` → `True` |

### 3. Logical operators
Used to combine conditions.

| Operator | Rule |
|----------|------|
| `and` | `True` only if **both** sides are true |
| `or` | `True` if **at least one** side is true |
| `not` | Reverses the truth value |

`and` / `or` use **short-circuit evaluation** and actually return one of their operands, not always `True`/`False`: `0 or 5` → `5`, `3 and 0` → `0`.

### 4. Assignment operators
Shortcuts for "update a variable using its own value".

| Operator | Same as | Operator | Same as |
|----------|---------|----------|---------|
| `x += 2` | `x = x + 2` | `x //= 2` | `x = x // 2` |
| `x -= 2` | `x = x - 2` | `x %= 3` | `x = x % 3` |
| `x *= 2` | `x = x * 2` | `x **= 2` | `x = x ** 2` |
| `x /= 2` | `x = x / 2` | | |

### 5. Bitwise operators
They work on the **binary digits** of integers. With `a = 12` (`1100`) and `b = 10` (`1010`):

| Operator | Name | Rule | Result |
|----------|------|------|--------|
| `a & b` | AND | 1 if both bits are 1 | `1000` = `8` |
| `a \| b` | OR | 1 if either bit is 1 | `1110` = `14` |
| `a ^ b` | XOR | 1 if the bits differ | `0110` = `6` |
| `~a` | NOT | flips all bits (`-(a+1)`) | `-13` |
| `a << 2` | Left shift | multiply by 2² | `48` |
| `a >> 2` | Right shift | integer-divide by 2² | `3` |

### 6. Identity operators: `is`, `is not`
They check whether two names point to the **same object in memory**, not just equal values.
- `==` compares **values**.
- `is` compares **identity**.

Small integers (about -5 to 256) are cached by Python, so `a = 2; b = 2; a is b` is `True`. Two separate lists with the same content are equal (`==`) but **not** identical (`is`).

### 7. Membership operators: `in`, `not in`
Check whether a value exists inside a sequence (string, list, tuple, set, dictionary keys).

### 8. Operator precedence
When an expression has several operators, Python evaluates them in this order (highest first). Use brackets `( )` to make your intention clear.

| Priority | Operators |
|----------|-----------|
| 1 | `( )` brackets |
| 2 | `**` |
| 3 | `+x`, `-x`, `~x` |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `<<`, `>>` |
| 7 | `&` |
| 8 | `^` |
| 9 | `\|` |
| 10 | `==`, `!=`, `<`, `>`, `<=`, `>=`, `is`, `in` |
| 11 | `not` |
| 12 | `and` |
| 13 | `or` |

### 9. Chained comparisons
Python lets you write `1 < x < 10` just like in mathematics. It is equivalent to `1 < x and x < 10`, but shorter and more readable.

````python
# 3. Print the truth tables for and, or, and not.
# AND: the result is True only when BOTH operands are True
# The format spec {value:<5} left-aligns each value in a 5-character wide column
print("--- AND ---")
print("A     B       | A and B")
print("-" * 21)
print(f"{True:<5} | {True:<5} | {True and True}")
print(f"{True:<5} | {False:<5} | {True and False}")
print(f"{False:<5} | {True:<5} | {False and True}")
print(f"{False:<5} | {False:<5} | {False and False}")

# OR: the result is True when AT LEAST ONE operand is True
print("\n--- OR ---")
print("A     B       | A or B")
print("-" * 20)
print(f"{True:<5} | {True:<5} | {True or True}")
print(f"{True:<5} | {False:<5} | {True or False}")
print(f"{False:<5} | {True:<5} | {False or True}")
print(f"{False:<5} | {False:<5} | {False or False}")

# NOT: reverses the boolean value (True becomes False and vice versa)
print("\n--- NOT ---")
print("A     | not A")
print("-" * 13)
print(f"{True:<5} | {not True}")
print(f"{False:<5} | {not False}")
````

````python
# 4. Start with x = 10 and apply +=, -=, *=, /=, //=, %=, **= in sequence, printing after each.
# Each augmented assignment updates x using its current value
# Expected values: 12, 10, 20, 10.0, 1.0, 0.0, 0.0
x = 10

# x = x + 2  -> 12
x += 2
print(x)
# x = x - 2  -> 10
x -= 2
print(x)
# x = x * 2  -> 20
x *= 2
print(x)
# x = x / 2  -> 10.0 (true division turns x into a float)
x /= 2
print(x)
# x = x % 3  -> 1.0 (remainder of 10.0 / 3)
x %= 3
print(x)
# x = x // 2 -> 0.0 (1.0 divided by 2, rounded down)
x //= 2
print(x)
# x = x ** 2 -> 0.0 (0.0 squared)
x **= 2
print(x)
````

````python
# 5. For a = 12, b = 10, print a & b, a | b, a ^ b, ~a, a << 2, and a >> 2.
# Explain each result in binary.

a = 12 # 0000 1100
b = 10 # 0000 1010

# AND: bit is 1 only if both bits are 1 -> 1100 & 1010 = 1000 (8)
print(f"a & b  = {a & b}")
# OR: bit is 1 if either bit is 1 -> 1100 | 1010 = 1110 (14)
print(f"a | b  = {a | b}")
# XOR: bit is 1 if the bits are different -> 1100 ^ 1010 = 0110 (6)
print(f"a ^ b  = {a ^ b}")
# NOT: flips every bit; for integers ~n = -(n + 1), so ~12 = -13
print(f"~a     = {~a}")
# Left shift by 2 = multiply by 2**2 -> 1100 becomes 110000 (48)
print(f"a << 2 = {a << 2}")
# Right shift by 2 = integer-divide by 2**2 -> 1100 becomes 11 (3)
print(f"a >> 2 = {a >> 2}")
````

````python
# 6. Show the difference between == and is using two lists and two small integers.
a = 2
b = 2
# '==' compares VALUES: 2 == 2 -> True
print(a==b)
# 'is' compares IDENTITY (same object in memory).
# Python caches small integers (about -5 to 256), so this is True.
# For lists: [1, 2] == [1, 2] is True, but [1, 2] is [1, 2] is False (two separate objects)
print(a is b)
````

````python
# 7. Use in and not in on a string and on a list.
# 'in' / 'not in' test whether a value is present in a sequence
a = [1,2,3]
# 1 is an element of the list -> True
print(1 in a)
# 'b' is not in the list -> False
print('b' in a)
# Opposite of 'in' -> False
print(1 not in a)
# 'b' is not in the list, so 'not in' is True
# Tip: 'in' also works on strings, e.g. 'py' in 'python' -> True
print('b' not in a)
````

````python
# 9. Check whether a number is even or odd using %, then again using the bitwise &.
num1 = 10
num2 = 9
# Method 1: remainder when divided by 2 -> 0 means even, 1 means odd
print(num1 % 2) # Even
print(num2 % 2) # Odd

# Method 2: bitwise AND with 1 checks the last bit -> 0 means even, 1 means odd
print(num1 & 1) # Even
print(num2 & 1) # Odd
````
